```python
import polars as pl
import gc

# --- CONFIGURATION ---
FILE_PATH = "data/capstone/kilt_dpr_100w_fixed.parquet"
CHUNK_SIZE = 5               # Read 5 rows at a time
MAX_CHUNKS_TO_PRINT = 3      # Stop after printing 3 chunks

def main():
    print(f"🔍 Scanning file: {FILE_PATH}...")
    
    try:
        # 1. Create a LazyFrame (does NOT load data into RAM)
        lf = pl.scan_parquet(FILE_PATH)
        
        # Get total rows safely
        total_rows = lf.select(pl.len()).collect().item()
        print(f"📊 Total rows in file: {total_rows:,}\n")
        print("-" * 50)
        
    except FileNotFoundError:
        print(f"❌ Could not find {FILE_PATH}. Check your path!")
        return

    offset = 0
    chunks_printed = 0
    
    # 2. Loop through the file in chunks
    while offset < total_rows:
        # Slice the exact rows we want, then pull ONLY those into RAM
        df_chunk = lf.slice(offset, CHUNK_SIZE).collect()
        
        if df_chunk.is_empty():
            break
            
        print(f"📦 --- CHUNK {chunks_printed + 1} (Offset: {offset}) ---")
        
        # Print the data (Polars automatically formats this nicely in the console)
        print(df_chunk.show())
        print(df_chunk["text"][0])
        print("\n")
        
        # Move our pointer forward
        offset += CHUNK_SIZE
        chunks_printed += 1
        
        # 3. Explicit Memory Cleanup
        del df_chunk
        gc.collect()
        
        # Stop early so we don't print millions of rows
        if chunks_printed >= MAX_CHUNKS_TO_PRINT:
            print("🛑 Reached max chunks to print. Stopping safely.")
            break

if __name__ == "__main__":
    main()
```
